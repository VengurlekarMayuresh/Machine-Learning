# Task A : Object Detection

```python
from google.colab import files
import os

print("Upload the 'main' image:")
main_upload = files.upload()
main_path = list(main_upload.keys())[0]

print("\nUpload the 'target' image (the template to find):")
target_upload = files.upload()
target_path = list(target_upload.keys())[0]

print(f"\nImages uploaded: Main={main_path}, Target={target_path}")
```

```python
import cv2
import numpy as np
from matplotlib import pyplot as plt

img_main = cv2.imread(main_path)
img_target = cv2.imread(target_path)

gray_main = cv2.cvtColor(img_main, cv2.COLOR_BGR2GRAY)
gray_target = cv2.cvtColor(img_target, cv2.COLOR_BGR2GRAY)

w, h = gray_target.shape[::-1]

# Template Matching
res = cv2.matchTemplate(gray_main, gray_target, cv2.TM_CCOEFF_NORMED)
min_val, max_val, min_loc, max_loc = cv2.minMaxLoc(res)

top_left = max_loc
bottom_right = (top_left[0] + w, top_left[1] + h)

cv2.rectangle(img_main, top_left, bottom_right, (0, 255, 0), 5)

plt.figure(figsize=(12, 8))
plt.subplot(121), plt.imshow(cv2.cvtColor(img_target, cv2.COLOR_BGR2RGB))
plt.title('Target'), plt.axis('off')
plt.subplot(122), plt.imshow(cv2.cvtColor(img_main, cv2.COLOR_BGR2RGB))
plt.title('Detected Location'), plt.axis('off')
plt.show()

print(f"Best match confidence: {max_val:.4f}")
```

# Task B : Review of Advanced Deep Learning-Based Object Detection Models

## Introduction
Object detection is a key task in computer vision that involves identifying and localizing objects within an image. Traditional methods relied on handcrafted features and sliding window techniques, which were computationally expensive and less accurate. Modern deep learning-based approaches have significantly improved performance by using convolutional neural networks (CNNs) to learn features automatically, resulting in higher accuracy and robustness.

---

## Models Studied

### 1. YOLO (You Only Look Once)
**Working Principle:**
- YOLO treats object detection as a single regression problem.
- The image is divided into a grid and each grid cell predicts bounding boxes and class probabilities.
- It processes the entire image in one forward pass, making it extremely fast.

**Advantages:**
- Very high speed (real-time detection).
- Good generalization across different domains.

**Limitations:**
- Slightly lower accuracy compared to region-based methods.
- Struggles with small object detection.

---

### 2. Faster R-CNN
**Working Principle:**
- Uses a two-stage approach:
  1. Region Proposal Network (RPN) generates candidate object regions.
  2. These regions are classified and refined.
- Shares convolutional layers between RPN and detection network.

**Advantages:**
- High accuracy and precise object localization.
- Performs well on small and complex objects.

**Limitations:**
- Slower compared to single-shot detectors.
- Computationally expensive.

---

### 3. SSD (Single Shot Detector)
**Working Principle:**
- Detects objects in a single pass like YOLO.
- Uses multiple feature maps at different scales for detecting objects of various sizes.
- Predicts bounding boxes and class scores simultaneously.

**Advantages:**
- Faster than Faster R-CNN.
- Better at detecting objects of different sizes compared to YOLO.

**Limitations:**
- Lower accuracy than Faster R-CNN.
- Performance drops for very small objects.

---

## Comparison of Models

| Model        | Speed        | Accuracy      | Best Use Case                  |
|--------------|-------------|--------------|--------------------------------|
| YOLO         | Very High    | Moderate      | Real-time applications         |
| Faster R-CNN | Low          | Very High     | High-precision tasks           |
| SSD          | High         | Moderate-High | Balanced speed and accuracy    |

---

## Real-World Applications

### 1. Surveillance
- Detecting suspicious activities and tracking individuals.
- YOLO is commonly used due to real-time performance.

### 2. Autonomous Driving
- Detecting pedestrians, vehicles and traffic signs.
- SSD and Faster R-CNN are preferred for better accuracy.

### 3. Healthcare
- Identifying tumors or anomalies in medical images.
- Faster R-CNN is widely used for precise detection.

---

## Conclusion
Deep learning-based object detection models have significantly outperformed traditional methods in both accuracy and efficiency. YOLO excels in speed, making it ideal for real-time systems. Faster R-CNN provides superior accuracy, suitable for critical applications. SSD offers a balance between speed and accuracy.

Each model has its own strengths and limitations and the choice depends on the specific application requirements. For real-time tasks, YOLO is preferred; for high-precision needs, Faster R-CNN is ideal; and for balanced performance, SSD is a good choice.
