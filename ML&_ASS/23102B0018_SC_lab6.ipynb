```python
def activation(net, threshold):

    if net >= threshold:

        return 1

    else:

        return 0

n = int(input("Enter number of inputs: "))

print("\nAssuming weight w = 1 for all inputs")

print("\nMcCulloch-Pitts Formula:")

print("Net = Σ (xi * wi)")

print("Output = 1 if Net >= Threshold else 0")

def generate_inputs(n):

    arr = []

    for i in range(2 ** n):

        binary = format(i, '0' + str(n) + 'b')

        arr.append([int(x) for x in binary])

    return arr

inputs = generate_inputs(n)

print("\nINPUT COMBINATIONS")

for row in inputs:

    print(row)

print("\n----- AND GATE -----")

threshold = n

print("Formula: Output = 1 if Σxi >=", threshold)

print("Truth Table:")

print("Inputs -> Output")

for row in inputs:

    net = sum(row)

    output = activation(net, threshold)

    print(row, "->", output)

print("\n----- OR GATE -----")

threshold = 1

print("Formula: Output = 1 if Σxi >=", threshold)

print("Truth Table:")

print("Inputs -> Output")

for row in inputs:

    net = sum(row)

    output = activation(net, threshold)

    print(row, "->", output)

print("\n----- XOR GATE -----")

print("Formula: Output = 1 if number of 1s is odd")

print("Truth Table:")

print("Inputs -> Output")

for row in inputs:

    output = sum(row) % 2

    print(row, "->", output)

print("\n----- NOT GATE -----")

print("Formula: Output = NOT x")

print("Truth Table:")

print("Input -> Output")

not_inputs = generate_inputs(1)

for row in not_inputs:

    output = 1 - row[0]

    print(row, "->", output)
```
